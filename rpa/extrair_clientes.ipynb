# Código-fonte RPA (Pyhton)

````python
import json
from bs4 import BeautifulSoup
import requests

# 1. Links dos arquivos no seu GitHub Pages
url_clientes = (
    'https://brunacunhahub.github.io/dio-lab-assistente-investimentos-rpa-n8n/'
)

# 2. Raspagem dos Clientes (RPA)
response = requests.get(url_clientes)
soup = BeautifulSoup(response.text, 'html.parser')

tabela = soup.find('table')
linhas = tabela.find_all('tr')[1:]  # Pula o cabeçalho

clientes = []
for linha in linhas:
  colunas = linha.find_all('td')
  if len(colunas) >= 4:
    cliente = {
        'nome': colunas[0].text.strip(),
        'email': colunas[1].text.strip(),
        'saldo': colunas[2].text.strip(),
        'perfil': colunas[3].text.strip(),
    }
    clientes.append(cliente)

# 3. URL do Webhook do N8N (Substitua pela URL gerada no seu N8N)
webhook_n8n_url = (
    'http://localhost:5678/webhook-test/investimentos-clientes'
)

# 4. Envio dos dados para o N8N
for c in clientes:
  res = requests.post(webhook_n8n_url, json=c)
  print(f"Cliente {c['nome']} enviado! Status: {res.status_code}")
````
